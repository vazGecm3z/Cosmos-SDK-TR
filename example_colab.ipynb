```python
import os
from huggingface_hub import HfFolder

# Mevcut oturumdaki tokenı al
token = HfFolder.get_token()

# Betiği oluştur
script_content = """
import os
os.environ["TORCH_COMPILE_DISABLE"] = "1"
os.environ["HUGGING_FACE_HUB_TOKEN"] = "{token}"
import torch
from diffusers import FluxPipeline
from model import T5EncoderModel, FluxTransformer2DModel

def run():
    quant_model_id = "HighCWu/FLUX.1-dev-4bit"
    model_id = "black-forest-labs/FLUX.1-dev"

    print("Modeller yukleniyor (HF Token ile)...")
    text_encoder_2 = T5EncoderModel.from_pretrained(quant_model_id, subfolder="text_encoder_2", torch_dtype=torch.bfloat16)
    transformer = FluxTransformer2DModel.from_pretrained(quant_model_id, subfolder="transformer", torch_dtype=torch.bfloat16)
    pipe = FluxPipeline.from_pretrained(model_id, text_encoder_2=text_encoder_2, transformer=transformer, torch_dtype=torch.bfloat16, token=os.environ["HUGGING_FACE_HUB_TOKEN"])

    pipe.to("cuda")

    # 1. Yöntem: Çoklu açı (Character Sheet) promptu
    prompt = (
        "Multiple-angle character sheet, split-view photo of the exact same 28-year-old Anatolian woman from different angles: "
        "front view, profile side view, and three-quarter view. Consistent facial identity, "
        "light tan olive skin, warm complexion, natural face without makeup, expressive brown eyes, gentle smile, "
        "simple cotton headscarf and knit cardigan, clean neutral background, soft studio light, 50mm lens, photorealistic"
    )
    print(f"Gorsel olusturuluyor: {{prompt}}")

    # 3 açının yan yana sığması için width=1024, height=512 yapıldı
    image = pipe(
        prompt,
        height=512,
        width=1024,
        guidance_scale=3.5,
        num_inference_steps=20,
    ).images[0]

    image.save("output_character_sheet.png")
    print("Gorsel basariyla kaydedildi: output_character_sheet.png")

if __name__ == "__main__":
    run()
""".format(token=token)

with open('run_pipeline.py', 'w') as f:
    f.write(script_content)

# Betiği çalıştır
!python3.10 run_pipeline.py

# Sonucu ekranda göster
from IPython.display import display
from PIL import Image
if os.path.exists('output_character_sheet.png'):
    display(Image.open('output_character_sheet.png'))
```
